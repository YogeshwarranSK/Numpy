Numpy - Numerical Python

Advantages :
1. Allows several mathematical calculations
2. Faster operations

```python
import numpy as np
```

```python
# Creating 1D array
a = np.array([1,2,3,4,5])
print(a)
```

```python
a.shape
```

```python
b = np.array([(1,2,3,4),(5,6,7,8)])
print(b)
```

```python
b.shape
```

```python
c = np.array([(1,2,3,4),(5,6,7,8)],dtype=float)
print(c)
```

Place Holders


```python
# Create array of zeroes
x = np.zeros((4,5))
print(x)
```

```python
# Create array of ones
y = np.ones((3,3))
print(y)
```

```python
# Create an array of particular value
z = np.full((5,4),5)
print(z)
```

```python
# Create an Identity matrix
a = np.eye(5)
print(a)
```

```python
# Create a numpy array with random values
b = np.random.random((3,4))
print(b)
```

```python
# Random integer values array
c = np.random.randint(10,100,(3,5))
print(c)
```

```python
# Array of evenly spaced values - specifying required values
d = np.linspace(10,40,5)
print(d)
```

```python
# Array of evenly spaced values - specifying the step
e = np.arange(10,40,5)
print(e)
```

```python
# Convert a list to numpy array
list2 = [10,20,30,40,50]
np_array = np.asarray(list2)
print(np_array)
type(np_array)

```

```python
tuple1 = (10,20,30,40,50)
np_array = np.asarray(tuple1)
print(np_array)
type(np_array)
```

Mathematical Operations

```python
list1 = [1,2,3,4,5]
list2 = [6,7,8,9,10]
print(list1 + list2) # Concatenate
print()

a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
print(a + b) # Element wise addition
print(a - b) # Element wise subtraction
print(a * b) # Element wise multiplication
print(a / b) # Element wise division
```

Array Manipulation


```python
a = np.random.randint(0,10,(2,3))
print(a)
print(a.shape)
```

```python
# transpose
b = np.transpose(a)
print(b)
print(b.shape)
```

```python
a = np.random.randint(0,10,(2,3))
print(a)
print(a.shape)
```

```python
# another method
b = a.T
print(b)
print(b.shape)
```

```python
# reshaping a array
a = np.random.randint(0,10,(2,3))
print(a)
print(a.shape)
```

```python
b = a.reshape(3,2)
print(b)
print(b.shape)
```
